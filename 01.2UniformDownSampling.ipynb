### Uniform Downsampling
Alternatively, use `uniform_down_sample` to downsample the point cloud by collecting every n-th points.

或者，可以使用`uniform_down_sample`函数，通过每隔 n 个点采样一次的方式，对点云进行下采样。

uniform: 每n个点保留一个  
voxel：将空间划分为体素网格，每个网格保留一个点，更适合点云配准等。  
相比于均匀下采样会得到更好的representation、更少的点（计算复杂度降低）

```python
import open3d as o3d
import numpy as np
import os
import sys
```

```python
print("Load a ply point cloud, print it, and render it")
pcd = o3d.io.read_point_cloud(r"C:\Users\Jin Cai\Desktop\open3D_tutorial\data\extract\DemoCropPointCloud\fragment.ply") 
o3d.visualization.draw_geometries([pcd])
```

```python
#每5个点取一个点
print("every 5th points are selected")
uni_down_pcd = pcd.uniform_down_sample(every_k_points= 5)
o3d.visualization.draw_geometries([uni_down_pcd])
```
