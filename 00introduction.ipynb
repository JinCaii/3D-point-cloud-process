```python
#读取点云文件
#io = input/output
#从指定路径读取fragment.ply点云文件，然后把这个点云对象保存到变量pcd中
pcd = o3d.io.read_point_cloud(r"C:\Users\User\Desktop\zivid_captures\000.ply")
print(pcd) #打印点云基本信息
#pcd.points保存的是pcd中每个点的三维坐标；np.asarray转换成array数组，每一行是一个点的三维坐标
print(np.asarray(pcd.points))
#给当前点云加一个坐标轴，不是必须
#红色x轴，绿色y轴，蓝色z轴。显示出的坐标抽是读取的ply文件里自带的，如摄像机拍出的照片，坐标轴就是相机坐标系的
coordinate_frame = o3d.geometry.TriangleMesh.create_coordinate_frame(
    size=0.5,
    origin=[0, 0, 0]
)
#打开open3D的可视化窗口，把pcd显示出来
#一个draw_geometries()可以同时显示多个几何对象，如o3d.visualization.draw_geometries([pcd1, pcd2])
o3d.visualization.draw_geometries([pcd, coordinate_frame])
```
